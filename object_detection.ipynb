```python
from IPython.display import display, Javascript
from google.colab.output import eval_js
from base64 import b64decode, b64encode
import cv2
import numpy as np
from ultralytics import YOLO

# 1. Load your trained model
model = YOLO("/content/runs/detect/coco_scratch_project/yolov8n_scratch-3/weights/best.pt")

# 2. JavaScript to handle webcam stream and keyboard quit listener
def webcam_stream():
    js = Javascript('''
        async function run() {
            // Create container for the live video stream
            const container = document.createElement('div');
            container.innerHTML = '<h3>Live YOLOv8 Object Detection (Press <b>Q</b> on your keyboard to stop)</h3>';
            document.body.appendChild(container);

            const img = document.createElement('img');
            img.style.display = 'block';
            container.appendChild(img);

            const video = document.createElement('video');
            video.style.display = 'none'; // Hidden video element to read frames

            const stream = await navigator.mediaDevices.getUserMedia({video: {width: 640, height: 480}});
            video.srcObject = stream;
            await video.play();

            const canvas = document.createElement('canvas');
            canvas.width = video.videoWidth;
            canvas.height = video.videoHeight;
            const ctx = canvas.getContext('2d');

            let isStreaming = true;

            // Listen for the 'q' or 'Q' keypress to stop the stream
            const handleKeyDown = (event) => {
                if (event.key === 'q' || event.key === 'Q') {
                    isStreaming = false;
                    stream.getTracks().forEach(track => track.stop()); // Turn off camera
                    container.remove(); // Remove UI container
                    window.removeEventListener('keydown', handleKeyDown);
                    console.log("Webcam stopped.");
                }
            };
            window.addEventListener('keydown', handleKeyDown);

            // Frame streaming loop
            while (isStreaming) {
                ctx.drawImage(video, 0, 0, canvas.width, canvas.height);
                const data = canvas.toDataURL('image/jpeg', 0.8);

                try {
                    // Send frame to Python and wait for processed result
                    const result = await google.colab.kernel.invokeFunction('notebook.SendFrame', [data], {});
                } catch (e) {
                    break; // Exit loop if cell is interrupted
                }

                await new Promise(resolve => setTimeout(resolve, 50)); // Frame pacing (~20 FPS)
            }

            // Cleanup in case loop breaks
            stream.getTracks().forEach(track => track.stop());
        }
        run();
    ''')
    display(js)

# Python callback to receive frame, run YOLOv8, and send it back to the browser
from google.colab import output
def handle_frame(data):
    try:
        header, encoded = data.split(',', 1)
        decoded_data = b64decode(encoded)
        nparr = np.frombuffer(decoded_data, np.uint8)
        img = cv2.imdecode(nparr, cv2.IMREAD_COLOR)

        # Run YOLOv8 object detection on the frame
        results = model(img, verbose=False)

        # Plot bounding boxes on frame
        annotated_frame = results[0].plot()

        # Encode back to JPEG to send to the browser UI
        _, encoded_img = cv2.imencode('.jpeg', annotated_frame, [int(cv2.IMWRITE_JPEG_QUALITY), 80])
        base64_img = b64encode(encoded_img).decode('utf-8')
        data_url = f"data:image/jpeg;base64,{base64_img}"

        # Update the image tag in JavaScript
        display(Javascript(f'''
            const imgs = document.getElementsByTagName('img');
            if (imgs.length > 0) {{
                imgs[imgs.length - 1].src = "{data_url}";
            }}
        '''))
    except Exception as e:
        pass

output.register_callback('notebook.SendFrame', handle_frame)

# Start streaming
print("Starting webcam live stream...")
webcam_stream()
```
